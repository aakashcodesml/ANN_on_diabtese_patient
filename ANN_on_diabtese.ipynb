```python
# scaling 
## we use scaling tecnique so that all  colums value comes in 0 to 1 so that is  not  giving any column higer weitage than other 
# fisrt split the data dependence and independence feature 
X= data.drop("Outcome",axis =1)
X
```

```python
y = data.Outcome
y
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
X                                                      # now all value comes between 0 to 1
```

```python
# train test split 
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.25,random_state = 0)
X_train
```

```python
y_train
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.Sequential([
    keras.layers.Dense(8,activation = 'relu',input_dim =(8)),
    keras.layers.Dense(1,activation = 'sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X_train, y_train, epochs=50)
```

# Evaluate the model

```python
model.evaluate(X_test,y_test)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred[:5]
```

```python
# now we understand like we define a thresold where where above the thresold becomes 1and below that becomes 0, let me create for loop
array =[]
for y in y_pred:
    if y>0.5:
        array.append(1)
    else:
        array.append(0)
```
